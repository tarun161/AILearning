```python
# Comprehensions in Python
def func(string: str) -> dict[str, int]:
    if(len(string)==0):
      raise ValueError("String is empty")
    words=string.split(" ")
    freq: dict[str,int] = {}
    for word in words :
      if len(word)>3 :
       freq[word]=freq.get(word,0)+1
    return freq


try:
  print(func("I Love coding"))
  func("")
except ValueError as e:
  print(e)



```

```python
# asyncio in Python
import asyncio
import time

async def fake_api_call(name: str, delay: float) -> str :
  await asyncio.sleep(delay)
  print(f"{name} finished after {delay}s")
  return f"{name} finished after {delay}s"

async def main() :
  start=time.perf_counter()
  results= await asyncio.gather(fake_api_call("First",1),fake_api_call("Second",2),fake_api_call("Third",3))
  print(results)
  print(time.perf_counter()-start)


await main()

```
