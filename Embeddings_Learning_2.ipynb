```python
import os
from langchain_openai import OpenAIEmbeddings
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from google.colab import userdata
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

client=embedding = OpenAIEmbeddings(openai_api_key=userdata.get('OPENAI_KEY'),
                             model="text-embedding-3-small"
                             )

text_inputs = [
    f"Title: {t['title']}. Description: {t['description']}. Category: {t['category']}"
    for t in tickets
]
print(len(tickets))
print(len(text_inputs))

embeddings = client.embed_documents(text_inputs)
print(len(embeddings))

# Generateing Embeddings

ticket_embeddings = np.array(embeddings)
print(ticket_embeddings.shape)

```

```python
# Similarity Scores


def show_similarities(query):
  print(f'query : {query}')

  embeddings = client.embed_query(query)

  query_embedding = np.array(embeddings)
  query_embedding_2d = query_embedding.reshape(1, -1)

  similarities = cosine_similarity(query_embedding_2d , ticket_embeddings)[0]


  best_match_idx = np.argmax(similarities)

  print('closest resolution(Highest cosine similarity) for the user query')

  print(tickets[best_match_idx]['resolution'])

  #to get top 5 closest ticet Ids

  top_indexes=np.argsort(similarities)[::-1][:5]

  for index in top_indexes:
     print(tickets[index]['title'])

```

```python
test_queries = [
    'Database is timing out',
    'Payment not working for foreign customers',
    'App crashes on iPhone',
    'Emails are not being sent'
]

for query in test_queries:
    print('**************')
    show_similarities(query)
    print('**************')


```
